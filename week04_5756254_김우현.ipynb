## 문제 1. Tensor 5개 생성과 속성 확인표 (40점)
서로 다른 Tensor 5개를 만들고, 아래 항목을 담은 표를 출력하세요.

| 이름 | shape | dim | dtype | device |

각 Tensor를 왜 그렇게 만들었는지 주석으로 한 줄 적으세요.


```python
# 답안
# 1. 서로 다른 Tensor 5개 생성
t1 = torch.tensor(3.14) # 스칼라 값 표현을 위한 0차원 텐서 생성
t2 = torch.tensor([1, 2, 3, 4, 5]) # 1차원 시퀀스 데이터를 표현하기 위한 정수형 텐서 생성
t3 = torch.zeros((3, 3), dtype = torch.float32) # 3x3 영행렬 초기화를 위한 2차원 텐서 생성
t4 = torch.ones((2, 3, 4), dtype = torch.int32) # 3차원 배열 데이터 구조 생성을 위한 1채움 텐서 생성
t5 = torch.randn((1, 3, 224, 224)) # 가로 세로 224 픽셀짜리 컬러 사진 1장의 형태를 만들기 위한 랜덤 텐서 생성

tensors = {
    "t1 (Scalar)" : t1,
    "t2 (Vector)" : t2,
    "t3 (Zero 2D)" : t3,
    "t4 (Ones 3D)" : t4,
    "t5 (Random Image)" : t5
}

# 2. 속성 확인표 데이터를 담을 리스트 작성
data = []
for name, t in tensors.items():
  data.append({
        "이름" : name,
        "shape" : list(t.shape),
        "dim" : t.dim(),
        "dtype" : t.dtype,
        "device" : t.device
  })

  # 3. Pandas DataFrame을 활용하여 표 형태로 출력
  df = pd.DataFrame(data)
  print(df.to_markdown(index = False))
```

## 문제 2. Tensor 연산 (20점)
(3, 4) 와 (4, 2) Tensor를 만들어 아래를 출력하세요.
- 행렬곱 결과와 그 shape
- 첫 번째 Tensor의 열별 평균과 행별 합계


```python
# 답안
# 1. (3, 4)와 (4, 2) 크기의 Tensor 생성
tensor1 = torch.randn(3, 4)
tensor2 = torch.randn(4, 2)

# 2. 행렬곱 연산 수행
result = torch.matmul(tensor1, tensor2)

# 3. 첫 번째 Tensor의 열별 평균 및 행별 합계
col_mean = tensor1.mean(dim = 0)
row_sum = tensor1.sum(dim = 1)

# 4. 결과 출력
print("1. 행렬곱 결과와 그 shape")
print("행렬곱 결과 :\n", result)
print("행렬곱 결과 shape : ", result.shape)
print("2. 첫 번째 Tensor (3, 4) 관련 연산")
print("열별 평균 (dim = 0) : ", col_mean)
print("행별 합계 (dim = 1) : ", row_sum)
```

## 문제 3. CPU와 GPU 비교 (20점)
같은 크기의 행렬곱을 CPU와 GPU에서 각각 실행해 걸린 시간을 출력하고, 차이를 한두 줄로 설명하세요.
GPU를 쓸 수 없는 경우에는 CPU 결과와 그 이유를 적으세요.


```python
# 답안
import time

# 1. 행렬 생성
A = torch.randn(3000, 3000)
B = torch.randn(3000, 3000)

# 2. CPU 연산 및 시간 측정
start = time.time()
result_cpu = A @ B
print("CPU 실행 시간:", time.time() - start, "초")

# 3. GPU 연산 및 시간 측정 (GPU 미지원 시 예외 처리)
if torch.cuda.is_available():
    A_gpu = A.to("cuda")
    B_gpu = B.to("cuda")

    start = time.time()
    result_gpu = A_gpu @ B_gpu
    print("GPU 실행 시간:", time.time() - start, "초")
else:
    print("GPU를 사용할 수 없는 환경입니다.")

# GPU는 수천 개의 코어로 대규모 행렬곱을 병렬로 동시에 계산하여, CPU에 비해 연산 속도가 훨씬 빠르다.

# 현재 실행 환경에 NVIDIA GPU 장치가 없거나,
# CUDA 드라이버 및 GPU용 PyTorch 라이브러리가 설정되어 있지 않기 때문입니다.
```

## 문제 4. 자동미분 (20점)
아래 두 가지를 하세요.
1. $y = 2x^3 + 5x$ 에서 $x=3$ 일 때의 도함수를 자동미분으로 구하고, 손으로 구한 값과 비교하세요.
2. 자기가 만든 수식 하나에 대해 미분값을 구하고, 손 계산과 일치하는지 확인하세요.


```python
# 답안

# 1. y = 2x^3 + 5x 에서 x = 3 일 때의 미분값
x1 = torch.tensor(3.0, requires_grad=True)
y1 = 2 * (x1**3) + 5 * x1
y1.backward()

auto_grad1 = x1.grad.item()
hand_grad1 = 6 * (3.0**2) + 5  # 손 계산 도함수: y' = 6x^2 + 5 -> 6*(3^2) + 5 = 59

print("1번 문제")
print(f"자동미분 결과: {auto_grad1}")
print(f"손 계산 결과  : {hand_grad1}")
print(f"일치 여부      : {auto_grad1 == hand_grad1}\n")

# 2. 자체 임의 수식: y = 3x^2 + 4x + 1 에서 x = 2 일 때의 미분값
x2 = torch.tensor(2.0, requires_grad=True)
y2 = 3 * (x2**2) + 4 * x2 + 1
y2.backward()

auto_grad2 = x2.grad.item()
hand_grad2 = 6 * 2.0 + 4  # 손 계산 도함수: y' = 6x + 4 -> 6*(2) + 4 = 16

print("2번 문제")
print(f"자동미분 결과: {auto_grad2}")
print(f"손 계산 결과  : {hand_grad2}")
print(f"일치 여부      : {auto_grad2 == hand_grad2}")
```
